# Script for extracting the masks presented in the paper

### Imports

```python
from src.utils.io_handler import load_mask_batch_from_disk
import matplotlib.pyplot as plt
```

### Appendix figures

```python
base_type="Diffusion"
specific_type="stepwise_guiding"
batch_size = 24

path = f"/home/leon/final_results_topo/{base_type}/steps_30/{specific_type}/seed_0/"

mask_batch, label_batch, num_masks = load_mask_batch_from_disk(path, batch_size, 0)

fig, axes = plt.subplots(4, 6, figsize=(15, 10))

for i, ax in enumerate(axes.flat):
    if i < len(mask_batch):
        ax.imshow(mask_batch[i].squeeze(), cmap='gray')

        ax.text(
            0.98, 0.98, str(label_batch[i].item()),
            color='red',
            fontsize=14,
            weight='bold',
            ha='right',
            va='top',
            transform=ax.transAxes
        )
    ax.axis('off')

plt.tight_layout()
plt.savefig(f"{base_type}_{specific_type}_30steps.pdf", bbox_inches='tight')
plt.show()
```
